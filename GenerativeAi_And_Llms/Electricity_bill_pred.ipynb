```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.ar_model import AutoReg
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.preprocessing import LabelEncoder

from sklearn.metrics import mean_absolute_error, r2_score
```

```python
df = pd.read_csv("D:\GenerativeAi_And_Llms\Datasets\electricity_bill_dataset.csv")
```

```python
df.head()
```

```python
#convert month to determine
df["Month"] = pd.to_datetime(df["Month"])

#sort for time series
df = df.sort_values("Month")
df.set_index("Month", inplace=True)
```

```python
encoder = LabelEncoder()
df["City"] = encoder.fit_transform(df["City"])
df["Company"] = encoder.fit_transform(df["Company"])
```

```python
ts = df["MonthlyHours"]

plt.figure(figsize=(10,5))
plt.plot(ts)
plt.title("Monthly Energy Usage Trend")
plt.xlabel("Month")
plt.ylabel("Usage (Hours)")
plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller

def adf_test(series):
    result = adfuller(series)
    print('ADF Statistic:', result[0])
    print('p-value:', result[1])
    print("stationarity" if result[1] <= 0.05 else "non-stationarity")

adf_test(ts)
```

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

plt.figure(figsize = (12, 5))
plt.subplot(121)
plot_acf(ts, ax=plt.gca(), lags=20)

plt.subplot(122)
plot_pacf(ts, ax=plt.gca(), lags=20)
plt.show()
```

```python
train_size = int(len(ts) * 0.8)
train, test = ts[:train_size], ts[train_size:]

ar_model = AutoReg(train, lags=5)
ar_fit = ar_model.fit()
ar_pred = ar_fit.predict(start=len(train), end=len(ts) - 1)
```

```python
from statsmodels.tsa.arima.model import ARIMA

arma_model = ARIMA(train, order=(3,0,2))
arma_fit = arma_model.fit()
arma_pred = arma_fit.predict(start=len(train), end=len(ts) - 1)
```

```python
arima_model = ARIMA(train, order=(3,1,2))
arima_fit = arima_model.fit()
arima_pred = arima_fit.predict(start=len(train), end=len(ts) - 1)
```

```python
plt.figure(figsize=(12,6))
plt.plot(test.index, test, label='Train')
plt.plot(test.index, ar_pred, label='Actual')
```

```python
plt.plot(test.index, ar_pred, label='AR')
plt.plot(test.index, arma_pred, label='ARMA')
plt.plot(test.index, arima_pred, label='ARIMA')
```

```python
plt.title("Monthly Energy Usage Forecast Prediction")
plt.xlabel("Month")
plt.ylabel("Usage (Hours)")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(train.index, train, label='Train')
plt.plot(test.index, test, label='Actual')
plt.plot(test.index, ar_pred, label='AR Predictions')
plt.plot(test.index, arma_pred, label='ARMA Predictions')   
plt.plot(test.index, arima_pred, label='ARIMA Predictions')
plt.title("Monthly Energy Usage Forecast Comparison")
plt.xlabel("Month")     
plt.ylabel("Monthly Hours")
plt.legend()
plt.show()
```

```python
target = "ElectricityBill"

exog_features = [
    "Fan", "Refrigerator", "AirConditioner", "Television", 
    "Monitor", "MotorPump", "MonthlyHours", 
    "TariffRate", "City", "Company"

]

X = df[exog_features]
y = df[target]
```

```python
train_size = int(len(ts) * 0.8)

x_train, x_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]

ar_model = AutoReg(train, lag=5)
ar_fit = ar_model.fit()
ar_pred = ar_fit.predict(start=len(train), end=len(ts) - 1)
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

model = SARIMAX(
    y_train,
    exog = x_train,
    order = (2,1,2),
    seasonal_order=(1,1,1,12)
)

model_fit = model.fit(disp=False)
```

```python
residuals = model_fit.resid

plt.figure(figsize=(12,5))
plt.subplot(121)
plt.plot(residuals)
plt.title("Residuals")

plt.subplot(122)
plt.hist(residuals, bins=30)
plt.title("Residuals Distribution")
plt.show()
```

```python
y_pred = model_fit.predict(
    start=len(y_train), 
    end=len(y) - 1, 
    exog=x_test
)
```

```python
plt.figure(figsize=(12,6))
plt.plot(y_train.index, y_train, label='Training Data')
plt.plot(y_test.index, y_test, label='Actual Bill')
plt.plot(y_test.index, y_pred, label='Predicted Bill')

plt.title("Electricity Bill Forecasting using Sarimax Model")
plt.xlabel("Month")
plt.ylabel("Electricity Bill")
plt.legend()
plt.show()
```
